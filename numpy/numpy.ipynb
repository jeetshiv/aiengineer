```python
import numpy as np
a=np.zeros((2,3))
b=np.ones((3,3))
c=np.full((2,2),7)
print("zero",a)
print("one",b)
print("full",c)
```

```python
import numpy as np
a=np.array([5,6,7])
b=np.array([2,2,2])
c=np.add(a,b)
d=np.subtract(a,b)
e=np.multiply(a,b)
f=np.divide(a,b)
g=np.power(a,b)
h=np.mod(a,b)
print("add=",c)
print("sub=",d)
print("mul=",e)
print("div=",f)
print("pow=",g)
print("mod=",h)
```
